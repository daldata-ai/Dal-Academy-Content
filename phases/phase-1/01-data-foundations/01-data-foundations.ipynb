```python
import json
import re
import sqlite3
import time
from collections import Counter, defaultdict, deque

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', 50)
pd.set_option('display.float_format', lambda x: f'{x:,.2f}')
plt.rcParams['figure.figsize'] = (9, 4)

# Dal visual system — aligned with daldata.ai branding
DAL_PRIMARY = '#B91C4B'
DAL_DARK = '#7A0F31'
DAL_DEEP = '#4D091E'
DAL_ROSE = '#F8E9EE'
DAL_INK = '#171418'
plt.rcParams.update({
    'figure.figsize': (9, 4),
    'axes.prop_cycle': plt.cycler(color=[DAL_PRIMARY, DAL_DARK, '#D75D7E', '#8F5368', '#D8A0B0']),
    'axes.titleweight': 'bold',
    'axes.titlecolor': DAL_DEEP,
    'axes.labelcolor': DAL_INK,
    'axes.edgecolor': '#D7C5CB',
    'grid.color': '#EADDE1',
    'grid.alpha': 0.65,
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
})

SEED = 42
rng = np.random.default_rng(SEED)
print('DAL | دال learning environment ready')
```

```python
# Reproducible synthetic commerce dataset used throughout the daldata.ai series.
rng = np.random.default_rng(SEED)

products = {
    'Laptop': ('Electronics', 900),
    'Phone': ('Electronics', 600),
    'Headphones': ('Electronics', 85),
    'Monitor': ('Electronics', 260),
    'Desk': ('Furniture', 320),
    'Chair': ('Furniture', 180),
    'Lamp': ('Furniture', 55),
    'Notebook': ('Stationery', 9),
    'Pen Set': ('Stationery', 12),
    'Backpack': ('Accessories', 70),
}

cities = ['Riyadh', 'Jeddah', 'Dammam', 'Makkah', 'Madinah']
segments = ['Consumer', 'SMB', 'Enterprise']
channels = ['web', 'mobile', 'partner']

customers = pd.DataFrame({
    'customer_id': np.arange(1, 301),
    'city': rng.choice(cities, 300, p=[0.34, 0.26, 0.16, 0.13, 0.11]),
    'segment': rng.choice(segments, 300, p=[0.58, 0.30, 0.12]),
    'signup_date': pd.to_datetime('2025-01-01') + pd.to_timedelta(rng.integers(0, 365, 300), unit='D')
})

n = 6500
product_names = rng.choice(list(products), n)
base_prices = np.array([products[p][1] for p in product_names], dtype=float)
orders = pd.DataFrame({
    'order_id': np.arange(100001, 100001 + n),
    'customer_id': rng.integers(1, 301, n),
    'order_ts': pd.to_datetime('2026-01-01') + pd.to_timedelta(rng.integers(0, 180*24*60, n), unit='m'),
    'product': product_names,
    'quantity': rng.integers(1, 6, n),
    'channel': rng.choice(channels, n, p=[0.48, 0.42, 0.10]),
    'status': rng.choice(['completed','cancelled','returned'], n, p=[0.84,0.10,0.06]),
})
orders['category'] = orders['product'].map({k:v[0] for k,v in products.items()})
orders['unit_price'] = np.round(base_prices * rng.normal(1.0, 0.07, n), 2)
orders['discount_pct'] = rng.choice([0, 0.05, 0.10, 0.15, 0.20], n, p=[0.50,0.18,0.17,0.10,0.05])

# Inject realistic quality issues on purpose.
missing_idx = rng.choice(orders.index, 90, replace=False)
orders.loc[missing_idx, 'unit_price'] = np.nan
orders = pd.concat([orders, orders.sample(45, random_state=SEED)], ignore_index=True)

# Add a few invalid-looking values for validation labs.
orders.loc[rng.choice(orders.index, 8, replace=False), 'quantity'] = 0

print(f'orders: {orders.shape[0]:,} rows | customers: {customers.shape[0]:,} rows')
orders.head(3)
```

| **Semi-structured** | self-describing nested records | JSON, logs, events | schema drift and flattening |

A structured table often lands in a warehouse, but the same structured data can also live in a lake or lakehouse. Semi-structured and unstructured data are natural fits for lake-style storage, while lakehouses add table reliability and SQL-friendly semantics over that flexible storage.

### Storage architecture vocabulary
- **Data warehouse:** curated analytical tables optimized for SQL, BI, dimensional models, and governed reporting.
- **Data lake:** scalable storage for raw, structured, semi-structured, and unstructured data—often built on object storage.
- **Data lakehouse:** combines lake-style open storage with warehouse-like table management, transactions, schema enforcement, and analytical query capabilities.

### Measurement scale matters

A column stored as an integer can still be an **identifier** rather than a measurable quantity. A category encoded as `1,2,3` is not automatically ordinal. Separate *storage type* from *semantic type*.

| Semantic scale | Example | Safe operations | Common mistake |
|---|---|---|---|
| Nominal | city, product | equality, counts, grouping | averaging encoded labels |
| Ordinal | satisfaction 1–5 | ordering, median | assuming equal spacing |
| Interval | Celsius | differences | treating zero as absence |
| Ratio | revenue, quantity | all arithmetic, ratios | ignoring units/currency |

```python
semantic_catalog = pd.DataFrame([
    ('order_id','identifier','nominal','unique order key'),
    ('customer_id','identifier','nominal','foreign key to customer'),
    ('order_ts','datetime','temporal','event timestamp'),
    ('quantity','integer','ratio','units purchased'),
    ('unit_price','decimal-like float','ratio','SAR per unit'),
    ('status','category','nominal','business state'),
], columns=['field','physical_representation','semantic_type','meaning'])
semantic_catalog
```

## 3. Semi-structured data: JSON and schema drift

```python
api_events = [
    {'event_id':'e1','customer':{'id':7,'city':'Riyadh'},'items':[{'sku':'Laptop','qty':1,'price':899.0}], 'source':'mobile'},
    {'event_id':'e2','customer':{'id':8,'city':'Jeddah'},'items':[{'sku':'Pen Set','qty':2,'price':11.5},{'sku':'Notebook','qty':3,'price':8.5}]},
    {'event_id':'e3','customer':{'id':9},'items':[], 'source':'web', 'campaign':'ramadan'},
]
print(json.dumps(api_events, indent=2))
```

A JSON record can add or omit keys without changing the file format. That flexibility is useful operationally, but analytics pipelines must decide how to handle absent fields, new fields, nested arrays, and version changes.

```python
flat_items = pd.json_normalize(
    api_events,
    record_path='items',
    meta=['event_id', ['customer','id'], ['customer','city'], 'source'],
    errors='ignore'
)
flat_items
```

## 4. Unstructured data becomes analyzable through representations

```python
reviews = [
    'Fast delivery and excellent laptop performance.',
    'Chair arrived damaged and support was slow.',
    'Great value. The desk was easy to assemble.',
    'Packaging was fine, but the headphones felt cheap.'
]

def tokenize(text):
    return re.findall(r"[a-z']+", text.lower())

word_counts = Counter(w for review in reviews for w in tokenize(review))
word_counts.most_common(10)
```

Text is unstructured, but token counts, TF-IDF vectors, embeddings, sentiment scores, and extracted entities turn it into numerical or categorical features. The representation should match the task—not every task needs a large model.

## 6. Data profiling: never analyze first and inspect later

```python
def profile_table(df):
    rows=[]
    for col in df.columns:
        s=df[col]
        rows.append({
            'column':col,
            'dtype':str(s.dtype),
            'null_pct':round(s.isna().mean()*100,2),
            'n_unique':s.nunique(dropna=True),
            'example':next((x for x in s.dropna().astype(str).head(1)), None)
        })
    return pd.DataFrame(rows)

profile_table(orders)
```

```python
quality_summary = {
    'rows': len(orders),
    'duplicate_rows': int(orders.duplicated().sum()),
    'missing_unit_price': int(orders['unit_price'].isna().sum()),
    'nonpositive_quantity': int((orders['quantity'] <= 0).sum()),
    'duplicate_order_ids': int(orders['order_id'].duplicated().sum())
}
quality_summary
```

## 7. Data quality dimensions

**Completeness** asks whether required data exists. **Validity** asks whether values satisfy domain rules. **Uniqueness** protects keys. **Consistency** checks agreement across fields/systems. **Timeliness** asks whether data arrives soon enough. **Accuracy** is hardest: a value can be syntactically valid and still be wrong.

```python
# Lightweight contract-like checks. In a production system, these become tests/monitors.
rules = {
    'order_id_unique': orders['order_id'].is_unique,
    'customer_id_present': orders['customer_id'].notna().all(),
    'status_allowed': orders['status'].isin({'completed','cancelled','returned'}).all(),
    'quantity_positive': (orders['quantity'] > 0).all(),
    'price_nonnegative_when_present': orders['unit_price'].dropna().ge(0).all(),
}
pd.Series(rules, name='passes')
```

## 8. Schema evolution: additive vs breaking changes

**Usually safer:** adding an optional field.  
**Potentially breaking:** renaming a field, changing units, changing a categorical meaning, changing precision, or repurposing an identifier.  

A professional pipeline versions contracts, validates producers and consumers, and documents business semantics alongside physical schema.
